````python
# Create the Multi-Agent Deep Q-Network implementation
class DQNAgent:
    """Deep Q-Network agent for network reconfiguration"""
    
    def __init__(self, state_size, action_size, agent_id, learning_rate=0.001):
        self.state_size = state_size
        self.action_size = action_size
        self.agent_id = agent_id
        self.learning_rate = learning_rate
        self.epsilon = 1.0
        self.epsilon_min = 0.01
        self.epsilon_decay = 0.995
        self.memory = deque(maxlen=10000)
        self.batch_size = 32
        
        # Neural network weights (simplified representation)
        self.q_network = self._build_network()
        self.target_network = self._build_network()
        self.update_target_network()
        
    def _build_network(self):
        """Build a simple neural network representation"""
        # In a full implementation, this would use PyTorch or TensorFlow
        # Here we simulate with weight matrices
        network = {
            'layer_1': np.random.normal(0, 0.1, (self.state_size, 128)),
            'bias_1': np.zeros(128),
            'layer_2': np.random.normal(0, 0.1, (128, 64)),
            'bias_2': np.zeros(64),
            'layer_3': np.random.normal(0, 0.1, (64, self.action_size)),
            'bias_3': np.zeros(self.action_size)
        }
        return network
    
    def _forward(self, state, network):
        """Forward pass through the network"""
        x = state
        
        # Layer 1
        x = np.dot(x, network['layer_1']) + network['bias_1']
        x = np.maximum(0, x)  # ReLU activation
        
        # Layer 2  
        x = np.dot(x, network['layer_2']) + network['bias_2']
        x = np.maximum(0, x)  # ReLU activation
        
        # Output layer
        x = np.dot(x, network['layer_3']) + network['bias_3']
        
        return x
    
    def predict(self, state):
        """Predict Q-values for given state"""
        return self._forward(state, self.q_network)
    
    def predict_target(self, state):
        """Predict Q-values using target network"""
        return self._forward(state, self.target_network)
    
    def remember(self, state, action, reward, next_state, done):
        """Store experience in replay memory"""
        self.memory.append((state, action, reward, next_state, done))
    
    def act(self, state):
        """Choose action using epsilon-greedy policy"""
        if np.random.random() <= self.epsilon:
            return random.randrange(self.action_size)
        
        q_values = self.predict(state)
        return np.argmax(q_values)
    
    def replay(self):
        """Train the network on a batch of experiences"""
        if len(self.memory) < self.batch_size:
            return
        
        # Sample batch from memory
        batch = random.sample(self.memory, self.batch_size)
        
        # Simplified training (in practice would use backpropagation)
        total_loss = 0
        for state, action, reward, next_state, done in batch:
            target = reward
            if not done:
                target = reward + 0.95 * np.max(self.predict_target(next_state))
            
            target_f = self.predict(state)
            target_f[action] = target
            
            # Simplified weight update (gradient descent)
            loss = (target_f[action] - self.predict(state)[action]) ** 2
            total_loss += loss
            
            # Update weights (simplified)
            self._update_weights(state, target_f, loss)
        
        # Decay epsilon
        if self.epsilon > self.epsilon_min:
            self.epsilon *= self.epsilon_decay
            
        return total_loss / self.batch_size
    
    def _update_weights(self, state, target, loss):
        """Simplified weight update"""
        # In practice, this would use proper backpropagation
        learning_rate = 0.001
        
        # Update output layer
        self.q_network['layer_3'] += learning_rate * np.outer(state, target - self.predict(state))
        
    def update_target_network(self):
        """Copy weights from main network to target network"""
        for key in self.q_network:
            self.target_network[key] = self.q_network[key].copy()

class MultiAgentDQN:
    """Multi-Agent Deep Q-Network system"""
    
    def __init__(self, network, num_agents=5):
        self.network = network
        self.num_agents = num_agents
        self.agents = []
        
        # Calculate state and action sizes
        state_size = len(network.get_state_vector())
        action_size = 2  # Open or close for each switch
        
        # Create agents
        for i in range(num_agents):
            agent = DQNAgent(state_size, action_size, i)
            self.agents.append(agent)
        
        self.training_history = {
            'episodes': [],
            'rewards': [],
            'losses': [],
            'epsilon': []
        }
    
    def get_joint_action(self, state):
        """Get joint action from all agents"""
        joint_action = []
        for agent in self.agents:
            action = agent.act(state)
            joint_action.append(action)
        return joint_action
    
    def train_episode(self):
        """Train agents for one episode"""
        state = self.network.get_state_vector()
        total_reward = 0
        total_loss = 0
        done = False
        step = 0
        max_steps = 50
        
        while not done and step < max_steps:
            # Get joint action
            joint_action = self.get_joint_action(state)
            
            # Apply action to network
            success, message = self.network.apply_switching_action(joint_action)
            
            # Calculate reward
            if success:
                pf_results = self.network.three_phase_power_flow()
                reward = self._calculate_reward(pf_results)
            else:
                reward = -100  # Penalty for invalid configuration
            
            # Get next state
            next_state = self.network.get_state_vector()
            
            # Check if done (improvement threshold met or max steps)
            done = (step >= max_steps) or (reward > 50)
            
            # Store experience and train each agent
            for i, agent in enumerate(self.agents):
                agent.remember(state, joint_action[i], reward, next_state, done)
                loss = agent.replay()
                if loss is not None:
                    total_loss += loss
            
            state = next_state
            total_reward += reward
            step += 1
        
        # Update target networks periodically
        if len(self.training_history['episodes']) % 10 == 0:
            for agent in self.agents:
                agent.update_target_network()
        
        # Record training history
        self.training_history['episodes'].append(len(self.training_history['episodes']))
        self.training_history['rewards'].append(total_reward)
        self.training_history['losses'].append(total_loss / len(self.agents))
        self.training_history['epsilon'].append(self.agents[0].epsilon)
        
        return total_reward, total_loss / len(self.agents)
    
    def _calculate_reward(self, pf_results):
        """Calculate reward based on power flow results"""
        # Multi-objective reward function
        loss_reduction_reward = -pf_results['power_losses'] * 10
        voltage_reward = -pf_results['voltage_deviation'] * 5
        
        # Penalty for voltage violations
        voltage_penalty = 0
        if pf_results['min_voltage'] < 0.95:
            voltage_penalty -= 50 * (0.95 - pf_results['min_voltage'])
        if pf_results['max_voltage'] > 1.05:
            voltage_penalty -= 50 * (pf_results['max_voltage'] - 1.05)
        
        total_reward = loss_reduction_reward + voltage_reward + voltage_penalty
        return total_reward
    
    def train(self, episodes=1000):
        """Train the multi-agent system"""
        print(f"Training MADRL system for {episodes} episodes...")
        
        for episode in range(episodes):
            reward, loss = self.train_episode()
            
            if episode % 100 == 0:
                print(f"Episode {episode}, Reward: {reward:.2f}, Loss: {loss:.4f}, "
                      f"Epsilon: {self.agents[0].epsilon:.3f}")
        
        print("Training completed!")
        return self.training_history

# Save the MADRL implementation
madrl_code = '''# Multi-Agent Deep Reinforcement Learning Module
# Implementation of DQN-based multi-agent system for network reconfiguration

import numpy as np
import random
from collections import deque

class DQNAgent:
    """Deep Q-Network agent for network reconfiguration"""
    
    def __init__(self, state_size, action_size, agent_id, learning_rate=0.001):
        self.state_size = state_size
        self.action_size = action_size
        self.agent_id = agent_id
        self.learning_rate = learning_rate
        self.epsilon = 1.0
        self.epsilon_min = 0.01
        self.epsilon_decay = 0.995
        self.memory = deque(maxlen=10000)
        self.batch_size = 32
        
        # Neural network weights (simplified representation)
        self.q_network = self._build_network()
        self.target_network = self._build_network()
        self.update_target_network()
        
    def _build_network(self):
        """Build a simple neural network representation"""
        network = {
            'layer_1': np.random.normal(0, 0.1, (self.state_size, 128)),
            'bias_1': np.zeros(128),
            'layer_2': np.random.normal(0, 0.1, (128, 64)),
            'bias_2': np.zeros(64),
            'layer_3': np.random.normal(0, 0.1, (64, self.action_size)),
            'bias_3': np.zeros(self.action_size)
        }
        return network
        
    # ... (rest of the implementation as above)

class MultiAgentDQN:
    """Multi-Agent Deep Q-Network system"""
    
    def __init__(self, network, num_agents=5):
        self.network = network
        self.num_agents = num_agents
        self.agents = []
        
        # Calculate state and action sizes
        state_size = len(network.get_state_vector())
        action_size = 2  # Open or close for each switch
        
        # Create agents
        for i in range(num_agents):
            agent = DQNAgent(state_size, action_size, i)
            self.agents.append(agent)
        
        self.training_history = {
            'episodes': [],
            'rewards': [],
            'losses': [],
            'epsilon': []
        }
    
    # ... (rest of the implementation as above)
'''

with open(f'{output_dir}/madrl_system.py', 'w') as f:
    f.write(madrl_code)

print("✓ Created Multi-Agent DQN implementation")
````

````python
# Create the Hybrid MADRL-GA Framework
class HybridMADRLGA:
    """Hybrid Multi-Agent Deep Reinforcement Learning with Genetic Algorithm Fine-tuning"""
    
    def __init__(self, network, num_agents=5):
        self.network = network
        self.madrl = MultiAgentDQN(network, num_agents)
        self.ga = GeneticAlgorithm(network, population_size=30, num_generations=50)
        
        self.performance_history = {
            'episodes': [],
            'rl_rewards': [],
            'ga_improvements': [],
            'total_improvements': [],
            'power_losses': [],
            'voltage_deviations': [],
            'execution_times': []
        }
    
    def train_rl_agent(self, episodes=500):
        """Train the MADRL agents"""
        print("Training MADRL agents...")
        training_history = self.madrl.train(episodes)
        return training_history
    
    def get_rl_solution(self):
        """Get current solution from trained RL agents"""
        state = self.network.get_state_vector()
        joint_action = self.madrl.get_joint_action(state)
        
        # Convert to chromosome format for GA
        rl_chromosome = [0] * len(self.ga.switchable_lines)
        for i, action in enumerate(joint_action[:len(rl_chromosome)]):
            rl_chromosome[i] = action
        
        return rl_chromosome
    
    def optimize_step(self, use_ga_fine_tuning=True):
        """Single optimization step using hybrid approach"""
        start_time = time.time()
        
        # Get RL solution
        rl_solution = self.get_rl_solution()
        
        # Apply RL solution and evaluate
        success, _ = self.network.apply_switching_action(rl_solution)
        if success:
            pf_results = self.network.three_phase_power_flow()
            rl_fitness = self.ga.evaluate_fitness(rl_solution)
        else:
            rl_fitness = -1000
            pf_results = {'power_losses': float('inf'), 'voltage_deviation': float('inf')}
        
        ga_improvement = 0
        final_solution = rl_solution
        
        if use_ga_fine_tuning and success:
            # Fine-tune with GA
            best_ga_solution, best_ga_fitness = self.ga.optimize(initial_solution=rl_solution)
            ga_improvement = best_ga_fitness - rl_fitness
            
            if ga_improvement > 0:
                final_solution = best_ga_solution
                # Apply final solution
                self.network.apply_switching_action(final_solution)
                pf_results = self.network.three_phase_power_flow()
        
        execution_time = time.time() - start_time
        
        # Record performance
        self.performance_history['episodes'].append(len(self.performance_history['episodes']))
        self.performance_history['rl_rewards'].append(rl_fitness)
        self.performance_history['ga_improvements'].append(ga_improvement)
        self.performance_history['total_improvements'].append(rl_fitness + ga_improvement)
        self.performance_history['power_losses'].append(pf_results['power_losses'])
        self.performance_history['voltage_deviations'].append(pf_results['voltage_deviation'])
        self.performance_history['execution_times'].append(execution_time)
        
        return {
            'rl_solution': rl_solution,
            'final_solution': final_solution,
            'rl_fitness': rl_fitness,
            'ga_improvement': ga_improvement,
            'power_losses': pf_results['power_losses'],
            'voltage_deviation': pf_results['voltage_deviation'],
            'execution_time': execution_time
        }
    
    def run_optimization(self, num_steps=10, use_ga_fine_tuning=True):
        """Run the complete hybrid optimization"""
        print(f"Running hybrid MADRL-GA optimization for {num_steps} steps...")
        
        results = []
        for step in range(num_steps):
            print(f"\\nStep {step + 1}/{num_steps}")
            result = self.optimize_step(use_ga_fine_tuning)
            results.append(result)
            
            print(f"RL Fitness: {result['rl_fitness']:.2f}")
            print(f"GA Improvement: {result['ga_improvement']:.2f}")
            print(f"Power Losses: {result['power_losses']:.4f} MW")
            print(f"Voltage Deviation: {result['voltage_deviation']:.4f}")
            print(f"Execution Time: {result['execution_time']:.2f} seconds")
        
        return results
    
    def generate_report(self):
        """Generate comprehensive performance report"""
        if not self.performance_history['episodes']:
            return "No optimization steps completed yet."
        
        report = f"""
HYBRID MADRL-GA OPTIMIZATION REPORT
=====================================

Network: {self.network.network_type}
Number of buses: {len(self.network.buses)}
Number of switches: {len(self.network.switches)}

PERFORMANCE SUMMARY:
-------------------
Total optimization steps: {len(self.performance_history['episodes'])}
Average RL fitness: {np.mean(self.performance_history['rl_rewards']):.2f}
Average GA improvement: {np.mean(self.performance_history['ga_improvements']):.2f}
Best total fitness: {max(self.performance_history['total_improvements']):.2f}

POWER SYSTEM METRICS:
--------------------
Initial power losses: {self.performance_history['power_losses'][0]:.4f} MW
Final power losses: {self.performance_history['power_losses'][-1]:.4f} MW
Loss reduction: {self.performance_history['power_losses'][0] - self.performance_history['power_losses'][-1]:.4f} MW
Loss reduction %: {((self.performance_history['power_losses'][0] - self.performance_history['power_losses'][-1]) / self.performance_history['power_losses'][0] * 100):.2f}%

Initial voltage deviation: {self.performance_history['voltage_deviations'][0]:.4f}
Final voltage deviation: {self.performance_history['voltage_deviations'][-1]:.4f}
Voltage improvement: {self.performance_history['voltage_deviations'][0] - self.performance_history['voltage_deviations'][-1]:.4f}

COMPUTATIONAL PERFORMANCE:
-------------------------
Average execution time per step: {np.mean(self.performance_history['execution_times']):.2f} seconds
Total execution time: {sum(self.performance_history['execution_times']):.2f} seconds
        """
        
        return report

# Create the main training and evaluation framework
class MADRLFramework:
    """Main framework for training and evaluating MADRL system"""
    
    def __init__(self):
        self.networks = {}
        self.hybrid_systems = {}
        self.results = {}
    
    def create_network(self, network_type='IEEE33'):
        """Create and store a network"""
        network = DistributionNetwork(network_type)
        self.networks[network_type] = network
        return network
    
    def create_hybrid_system(self, network_type='IEEE33', num_agents=5):
        """Create hybrid MADRL-GA system"""
        if network_type not in self.networks:
            self.create_network(network_type)
        
        network = self.networks[network_type]
        hybrid_system = HybridMADRLGA(network, num_agents)
        self.hybrid_systems[network_type] = hybrid_system
        return hybrid_system
    
    def run_comprehensive_evaluation(self):
        """Run comprehensive evaluation on both test systems"""
        results = {}
        
        for network_type in ['IEEE33', 'IEEE123']:
            print(f"\\n{'='*50}")
            print(f"EVALUATING {network_type} SYSTEM")
            print(f"{'='*50}")
            
            # Create hybrid system
            hybrid_system = self.create_hybrid_system(network_type, num_agents=5)
            
            # Train RL agents
            training_history = hybrid_system.train_rl_agent(episodes=200)
            
            # Run hybrid optimization
            optimization_results = hybrid_system.run_optimization(num_steps=5, use_ga_fine_tuning=True)
            
            # Generate report
            report = hybrid_system.generate_report()
            
            results[network_type] = {
                'training_history': training_history,
                'optimization_results': optimization_results,
                'report': report,
                'hybrid_system': hybrid_system
            }
        
        self.results = results
        return results

# Save the complete framework
framework_code = '''# Hybrid MADRL-GA Framework
# Complete framework for distribution network reconfiguration using
# Multi-Agent Deep Reinforcement Learning with Genetic Algorithm fine-tuning

import numpy as np
import time
from distribution_network import DistributionNetwork
from madrl_system import MultiAgentDQN
from genetic_algorithm import GeneticAlgorithm

class HybridMADRLGA:
    """Hybrid Multi-Agent Deep Reinforcement Learning with Genetic Algorithm Fine-tuning"""
    
    def __init__(self, network, num_agents=5):
        self.network = network
        self.madrl = MultiAgentDQN(network, num_agents)
        self.ga = GeneticAlgorithm(network, population_size=30, num_generations=50)
        
        self.performance_history = {
            'episodes': [],
            'rl_rewards': [],
            'ga_improvements': [],
            'total_improvements': [],
            'power_losses': [],
            'voltage_deviations': [],
            'execution_times': []
        }
    
    def train_rl_agent(self, episodes=500):
        """Train the MADRL agents"""
        print("Training MADRL agents...")
        training_history = self.madrl.train(episodes)
        return training_history
    
    def optimize_step(self, use_ga_fine_tuning=True):
        """Single optimization step using hybrid approach"""
        start_time = time.time()
        
        # Get RL solution
        state = self.network.get_state_vector()
        joint_action = self.madrl.get_joint_action(state)
        
        # Convert to chromosome format for GA
        rl_solution = [0] * len(self.ga.switchable_lines)
        for i, action in enumerate(joint_action[:len(rl_solution)]):
            rl_solution[i] = action
        
        # Apply RL solution and evaluate
        success, _ = self.network.apply_switching_action(rl_solution)
        if success:
            pf_results = self.network.three_phase_power_flow()
            rl_fitness = self.ga.evaluate_fitness(rl_solution)
        else:
            rl_fitness = -1000
            pf_results = {'power_losses': float('inf'), 'voltage_deviation': float('inf')}
        
        ga_improvement = 0
        final_solution = rl_solution
        
        if use_ga_fine_tuning and success:
            # Fine-tune with GA
            best_ga_solution, best_ga_fitness = self.ga.optimize(initial_solution=rl_solution)
            ga_improvement = best_ga_fitness - rl_fitness
            
            if ga_improvement > 0:
                final_solution = best_ga_solution
                self.network.apply_switching_action(final_solution)
                pf_results = self.network.three_phase_power_flow()
        
        execution_time = time.time() - start_time
        
        return {
            'rl_solution': rl_solution,
            'final_solution': final_solution,
            'rl_fitness': rl_fitness,
            'ga_improvement': ga_improvement,
            'power_losses': pf_results['power_losses'],
            'voltage_deviation': pf_results['voltage_deviation'],
            'execution_time': execution_time
        }
    
    # Additional methods for optimization and reporting...

class MADRLFramework:
    """Main framework for training and evaluating MADRL system"""
    
    def __init__(self):
        self.networks = {}
        self.hybrid_systems = {}
        self.results = {}
    
    def create_network(self, network_type='IEEE33'):
        """Create and store a network"""
        network = DistributionNetwork(network_type)
        self.networks[network_type] = network
        return network
    
    def create_hybrid_system(self, network_type='IEEE33', num_agents=5):
        """Create hybrid MADRL-GA system"""
        if network_type not in self.networks:
            self.create_network(network_type)
        
        network = self.networks[network_type]
        hybrid_system = HybridMADRLGA(network, num_agents)
        self.hybrid_systems[network_type] = hybrid_system
        return hybrid_system
    
    # Additional methods for comprehensive evaluation...
'''

with open(f'{output_dir}/hybrid_framework.py', 'w') as f:
    f.write(framework_code)

print("✓ Created Hybrid MADRL-GA Framework")
````
